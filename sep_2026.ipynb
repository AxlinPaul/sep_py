```python
class Student:
    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no


class Test(Student):
    def __init__(self, name, roll_no, tmark):
        super().__init__(name, roll_no)
        self.tmark = tmark


class Sports:
    def __init__(self, smark):
        self.smark = smark


class Grades(Test, Sports):
    def __init__(self, name, roll_no, tmark, smark):
        Test.__init__(self, name, roll_no, tmark)
        Sports.__init__(self, smark)

    def total_mark(self):
        self.tomark = self.tmark + self.smark
        return self.tomark

    def calgrade(self):
        total = self.total_mark()

        print("Name:", self.name)
        print("Roll No:", self.roll_no)
        print("Test Mark:", self.tmark)
        print("Sports Mark:", self.smark)
        print("Total Mark:", total)

        if total > 80:
            print("Grade A")
        elif total > 50:
            print("Grade B")
        else:
            print("Fail")



g1 = Grades("Axlin", 101, 45, 40)
g1.calgrade()
```
